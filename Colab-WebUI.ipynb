## Launch WebUI
## 启动 WebUI

```
%%writefile patch_jinja.py
import sys
import os
import jinja2

path = os.path.join(os.path.dirname(jinja2.__file__), 'utils.py')
print(f"Patching: {path}")

with open(path, 'r') as f:
    lines = f.readlines()

with open(path, 'w') as f:
    for line in lines:
        if 'except KeyError:' in line and 'self.missing(key)' in line:
            f.write(line.replace('except KeyError:', 'except (KeyError, TypeError):'))
        else:
            f.write(line)

print("Patch applied successfully.")
```
